```python
import pandas as pd

titanic = pd.read_csv("../exercice_08/data/titanic_propre.csv")
titanic.shape
```

```python
survie_par_classe_et_sexe = titanic.groupby(["class", "sex"])["survived"].mean().round(2)
survie_par_classe_et_sexe
```

```python
pd.pivot_table(titanic, values="survived", index="class", columns="sex", aggfunc="mean").round(2)
```

```python
titanic["tranche_age"] = pd.cut(
    titanic["age"], bins=[0, 12, 18, 60, 100],
    labels=["enfant", "ado", "adulte", "senior"],
)
survie_par_tranche_age = titanic.groupby("tranche_age")["survived"].mean().round(2)
survie_par_tranche_age
```

```python
prix_moyen_par_port = titanic.groupby("embark_town")["fare"].mean().round(2)
prix_moyen_par_port
```

```python
repartition_classes_par_port = pd.crosstab(
    titanic["embark_town"], titanic["class"], normalize="index"
).round(2)
repartition_classes_par_port
```

```python
survie_seul_ou_non = titanic.groupby("alone")["survived"].mean().round(2)
survie_seul_ou_non
```

```python
titanic["alone"] = (titanic["sibsp"] + titanic["parch"]) == 0
```
