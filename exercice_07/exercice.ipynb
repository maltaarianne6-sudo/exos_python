```python
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic')
```

```python
titanic.head()
```

```python
import pandas as pd
```

```python
nb_passengers_moins_de_18_ans = (titanic['age'] < 18).sum()
print("Nombre de passagers de moins de 18 ans :", nb_passengers_moins_de_18_ans)
```

```python
billets_les_plus_chers = titanic[["class", "embark_town", "fare"]].sort_values("fare", ascending=False).head(5)
print("Les 5 billets les plus chers :", billets_les_plus_chers)
```

```python
première_classe = titanic[titanic["class"] == "First"]
nb_passagers_premiere_classe = première_classe.shape[0]
print("Nombre de première classe :", nb_passagers_premiere_classe)
```

```python
part_des_femmes_premiere_classe = (première_classe["sex"] == "female").mean() * 100
print("Pourcentage de femmes de première classe :", part_des_femmes_premiere_classe)
```

```python
taux_de_survie_premiere_classe = titanic[(titanic["class"] == "First") & (titanic["sex"] == "female")]["survived"].mean() * 100
print("Taux de survie des femmes de première classe :", taux_de_survie_premiere_classe)
```

```python
taux_de_survie_troisieme_classe = titanic ["survived"][(titanic["class"] == "Third") & (titanic["sex"] == "male")].mean() * 100
print("Taux de survie des hommes de troisième classe :", taux_de_survie_troisieme_classe)
```
